```python
from PyPDF2 import PdfReader, PdfWriter
```

```python
### Open the current PDF

file_pdf=PdfReader('pdf_example.pdf')

## Object for PDF writer
out_pdf=PdfWriter()
```

```python
file_pdf # we see that is a object of PyPDF2 type
```

```python
len(file_pdf.pages) # see the length of the pdf
```

```python
type(file_pdf.pages) # we see that the pdf pages are stored a list of some kind
```

```python
for i in range(len(file_pdf.pages)):
    page_details=file_pdf.pages[i]
    ### adding pages to the output page
    out_pdf=out_pdf.add(page_details)
```
