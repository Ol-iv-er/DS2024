## SHA (Secure Hash Algorithms) - Base of BlockChain

- SHA256: This Hash function belongs to the class SHA-2, the internal block of it is 32 bits.
- SHA384: This hash fucntion belons to the hash class SHA-2, the internal block of it is 32 bits. This is one of the truncated version.
- SHA224: This hash fucntion belons to the hash class SHA-2, the internal block of it is 32 bits. This is one of the truncated version.
- SHA512: This hash fucntion belons to the hash class SHA-2, the internal block of it is 32 bits. This is one of the truncated version.
- SHA1: The 160 bit hash function that resembles MD5 had in working and was discontinued to be used seeing its security vulnerabilities.

```python
import hashlib
```

## SHA256

```python
str1='Oliver'
### Proccess the Hashing
### First step is to encode
### Then apply hashing Algorithms
encoded_str = str1.encode()
hashed_val = hashlib.sha256(encoded_str)
print(hashed_val)
```

```python
### convert this values to hexadecimal
hashed_val.hexdigest()
```

```python
len(hashed_val.hexdigest())
```

## SHA 384

```python
hashed_val1=hashlib.sha384(encoded_str)
hashed_val1.hexdigest()
```

```python
len(hashed_val1.hexdigest())
```

### SHA224

```python
hashed_val2=hashlib.sha224(encoded_str)
hashed_val2.hexdigest()
```

```python
len(hashed_val2.hexdigest())
```

### SHA512

```python
hashed_val3=hashlib.sha512(encoded_str)
hashed_val3.hexdigest()
```

```python
len(hashed_val3.hexdigest())
```

```python
hashed_val=hashlib.sha1(encoded_str)
hashed_val.hexdigest()
```

```python
len(hashed_val.hexdigest())
```
