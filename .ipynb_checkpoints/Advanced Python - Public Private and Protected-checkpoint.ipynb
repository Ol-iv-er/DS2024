## Python OOPS - Public, Private, and Protected

```python
### All class variables are pulbic
class Car():
    def __init__(self, engine_type, doors, windows):
        self.engine_type=engine_type
        self.doors=doors
        self.windows=windows
        
```

```python
audi=Car("Diesel", 4, 5)
```

```python
audi
```

```python
dir(audi) ## atributes, the ones that have _ are protected
```

```python
audi.windows
```

```python
## All the class variables are protected
class Car():
    def __init__(self, engine_type, doors, windows):
        self._engine_type=engine_type
        self._doors=doors
        self._windows=windows
        
```

```python
audi=Car("Diesel", 4, 5)
```

```python
# dir(audi)
```

```python
## Since these varaibels are protected we can access them here in the child class only
class Truck(Car):
    def __init__(self, engine_type, doors, windows, horse_power):
        super().__init__(engine_type, doors, windows)
        self.horse_power=horse_power
```

```python
truck=Truck("Diesel", 4, 5, 5000)
dir(truck) # we change the public to protects by adding an _ to varaiables, that lets the developers know what can be changed and what should not
```

```python
truck._doors=6 
```

```python
truck._doors
```

### Private

```python
class Car():
    def __init__(self, engine_type, doors, windows):
        self.__engine_type=engine_type
        self.__doors=doors
        self.__windows=windows
        
```

```python
audi=Car("Diesel", 4,5)
```

```python
dir(audi)
```

```python
#'_Car__doors', '_Car__engine_type', '_Car__windows' show us that we have private variables but this lets us know that it is private
```
