```python
lst1=['o', 'p', 'e']
lst2= [1,2,3]
```

```python
zip_object = zip(lst1, lst2)
```

```python
## zip object is iterable
next(zip_object)
```

```python
list(zip_object)
```

```python
even=[2,4,6,8]
odd=[1,3,5]
```

```python
list(zip(even,odd)) # will zip according to the smaller lengthed object
```

```python
zip_object = zip(lst1, lst2)
for i,j in zip_object:
    print(i,j)
```

```python
## what about three elements
lst3=['a', 'b', 'c']

new_zip = zip(lst1, lst2, lst3)
for i,j,k in new_zip:
    print(i,j,k)
```

```python
list(zip())
```

```python
dict1={'first_name': 'Oliver', 'last_name': 'Ramirez', 'Age':24}
dict2={'first_name': 'Harry', 'last_name': 'Gohs', 'Age':22}
```

```python
dictionary=zip(dict1.items(), dict2.items())
```

```python
list(dictionary)
```

```python
for (i,j), (i2,j2) in dictionary:
    print(i,j)
    print(i2,j2)
```
