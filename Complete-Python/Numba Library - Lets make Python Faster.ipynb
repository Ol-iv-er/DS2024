## Quick Guide to Numba

Numba is a just-in-time compiler for Python that workds best on code that uses NumPy arrays and functions, and loops. The most common way to use Numba is through it collection of decorators that can be applied to your function to instruct Numba to compile them. When a call is made to a Numba decoated function it is complied to machine code "just-in-time" for execution and all or part of your code can subsequently run at native machine code speed.

```python
from numba import jit
import numpy as np
import time

x=np.arange(100).reshape(10,10)

@jit(nopython=True)
def go_fast(a): # Function is complied and runs in machine code
    trace = 0.0
    for i in range(a.shape[0]):
        trace += np.tanh(a[i,i])
    return a + trace


# DO NOT REPORT THIS ... COMPLIATION TIME IS INCLUDED IN THE EXECUTION TIME!
start=time.time()
go_fast(x)
end=time.time()
print("Elapsed (with compilation) = %s" % (end - start))

# NOW THE FUNCTION SI COMPLIED, RE-TIME IT EXECUTION FROM CACHE
start=time.time()
go_fast(x)
end=time.time()
print("Elapsed (after compilation) = %s" % (end - start))

# NOW THE FUNCTION IS COMPLIED, RE-TIME IT EXECUTION FROM CACHE
start=time.time()
go_fast(x)
end=time.time()
print("Elapsed2 (after compilation) = %s" % (end - start))
```

Once complied the code will be stored as machine code on cache and called from cache every time after making for a faster compilation time.
