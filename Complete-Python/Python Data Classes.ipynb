## Python Data Classes

In Python, the dataclass decorator is a feature introduced in Python 3.7 that provides a concise way to define classes primarily to store data. It automatically generates several special methods, such as **init**, **repr**, and **eq**, based on the class attributes you define. This simplifies the process of creating and working with data-focused classes.

```python
class Person:
    def __init__(self, name, age, nationality):
        self.name = name
        self.age = age
        self.nationality = nationality
        
person1=Person('Oliver', 24, 'Mexican-American')
print(person1)
```

```python
from dataclasses import dataclass

@dataclass
class Person:
    name:str
    age:int
    occupation:str
        
person1=Person('Oliver', 24, 'unemployed')

print(person1)
```

The @dataclass decorator automatically generates the following methods for you:

1. __init( )__: Initializes the object and assigns the provided values to the attributes.
2. __repr( )__: Provides a string represention of the object.
3. __eq( )__: Implements equality comparison between two objects of the class based on their attributes.

```python
@dataclass
class Person:
    name:str
    age:int
    occupation:str = 'AI'
        
person1=Person('Oliver', 24)
print(person1)
```

```python
person1.occupation = 'ML Engineer'
print(person1)
```

### Imutable class

```python
@dataclass(frozen=True) # the class attributes cannot change
class Point:
    x:int
    y:int

point=Point(3,4)
print(point)
```

```python
point.x=3
```

## Inheritance

```python
@dataclass
class Person:
    name:str
    age:int
        
        
@dataclass
class Employee(Person):
    employee_id:str
    department:str
        
person=Person('Oliver', 24)

employee=Employee('Oliver', 24, '123', 'AI') # when inherited the values for the partent class are put first
```

```python
print(employee)
```

## Nested data class

```python
@dataclass
class Address:
    street:str
    city:str
    zip_code:str
        
@dataclass
class Person:
    name:str
    age:str
    address:Address
    
address = Address('Main St', 'Salinas', '93906')
person=Person('Oliver', 24, address)
```

```python
person
```

```python
person.address.street
```

```python
person.address.city
```

```python
person.address.zip_code
```
